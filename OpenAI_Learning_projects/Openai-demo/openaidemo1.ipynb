```python
prompt = "What is Python?"

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role":"system","content":"You are a helpful assistant."},
        {"role":"user","content":prompt}
    ],
    max_tokens=25
)

print(response.choices[0].message.content)
```

```python
prompt = "What is Python?"

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role":"system","content":"You are a helpful assistant."},
        {"role":"user","content":prompt}
    ],
    max_tokens=25
)

print(response)
```

```python
prompt = "What is Python?"

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role":"system","content":"You are a helpful assistant."},
        {"role":"user","content":prompt}
    ],
    temperature=0.6
)

print(response.choices[0].message.content)
```

```python
for i in range(3):
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role":"system","content":"You are a helpful assistant."},
            {"role":"user","content":"What is Python?"}
        ],
        temperature=0.8
    )

    print(f"Response {i+1}")
    print(response.choices[0].message.content)
    print()
```

```python
prompt = "Give me the sentiment of this sentence: This movie is amazing!"

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role":"system","content":"You are a helpful assistant."},
        {"role":"user","content":prompt}
    ]
)

print(response.choices[0].message.content)
```

```python
# ChatCompletion API 

prompt = "Give me hindi translation of this sentence: This Movie is Amazing!"

response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": prompt}
    ],
    temperature=0.6
)

print(response.choices[0].message.content)
```
